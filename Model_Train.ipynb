```python
import pandas as pd

# CSV ဖိုင်ဖတ်
df = pd.read_csv("questions.csv")

# အရင်ဆုံး dataset ကို စစ်ကြည့်
print(df.head())
```

```python
import random

def recommend_question(subject_input):
    filtered = df[df['Subject'].str.lower() == subject_input.lower()]
    if not filtered.empty:
        return random.choice(filtered['Question'].tolist())
    else:
        return "No questions found for this subject."
```
